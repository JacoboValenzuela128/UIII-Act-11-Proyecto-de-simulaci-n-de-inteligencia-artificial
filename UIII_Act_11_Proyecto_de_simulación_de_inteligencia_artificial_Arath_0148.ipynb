# AGRICULTOR

```python
# ==========================================================
# SIMULADOR IA: AGRO-IA SMART MANAGER (Funciones y Tablas)
# ==========================================================

# Estructuras de datos iniciales (vacías)
lista_agricultores = []
agricultor = {}

# ------------------------------------------------------
# FUNCIÓN 1: EVALUAR ESTRATEGIA (Lógica de IA con IF-ELIF)
# ------------------------------------------------------
def analizar_con_ia(cultivo, hectareas, fertilizante):
    print("\n[🤖 IA Analizando consistencia del suelo y recursos...]")

    # Regla de IA 1: Maíz con Químicos en grandes extensiones
    if cultivo.lower() == "maiz" and fertilizante.lower() == "quimico":
        return "⚠️ Riesgo de erosión alta. La IA sugiere rotar a leguminosas el próximo ciclo."

    # Regla de IA 2: Cultivos orgánicos sustentables
    elif fertilizante.lower() == "organico" and hectareas <= 10:
        return "🌱 ¡Estrategia Eco-Sustentable Perfecta! Ideal para certificación orgánica."

    # Regla de IA 3: Monocultivos masivos
    elif hectareas > 50:
        return "🚜 Alerta de Escala: Requiere monitoreo satelital por alta densidad de hectáreas."

    # Diagnóstico por defecto
    else:
        return "✅ Parámetros estables. El cultivo cumple con el estándar de rendimiento óptimo."

# ------------------------------------------------------
# FUNCIÓN 2: MOSTRAR TABLA DE REGISTROS (Ciclo FOR y Alineación)
# ------------------------------------------------------
def mostrar_tabla_agricultores():
    if lista_agricultores == []:
        print("\n❌ La base de datos está vacía. No hay agricultores registrados.")
    else:
        print("\n========================================================================================================================")
        print("                                       REPORTE GENERADO POR AGRO-IA (BASE DE DATOS)                                     ")
        print("========================================================================================================================")

        # Cabecera de la tabla con espaciados fijos para que quede perfecta
        # El formato {:<15} significa: alinea a la izquierda y reserva 15 espacios.
        print("{:<15} {:<12} {:<18} {:<12} {:<15} {:<15} {:<20}".format(
            "Cultivo", "Hectáreas", "Maquinaria", "Temporada", "Ubicación", "Fertilizante", "Diagnóstico IA"
        ))
        print("------------------------------------------------------------------------------------------------------------------------")

        # Ciclo FOR para imprimir cada fila de la tabla
        for agri in lista_agricultores:
            print("{:<15} {:<12} {:<18} {:<12} {:<15} {:<15} {:<20}".format(
                agri["Tipo de cultivo"],
                str(agri["hectáreas"]) + " ha", # Convertimos a texto para concatenar
                agri["maquinaría agrícola"],
                agri["temporada de cosecha"],
                agri["ubicación del terreno"],
                agri["fertilizantes usados"],
                agri["diagnostico_ia"]
            ))
        print("========================================================================================================================")


# ==========================================================
# CICLO PRINCIPAL (MENU INTERACTIVO)
# ==========================================================
while True:
    print("\n--- MENÚ DE CONTROL AGRO-IA ---")
    print("1. Registrar datos de un Agricultor (Entrada Key-Value)")
    print("2. Ver reporte de Agricultores en Tabla")
    print("3. Salir del sistema")

    opcion = input("\nSelecciona una opción (1-3): ")

    # ------------------------------------------------------
    # OPCIÓN 1: ENTRADA DE DATOS (Diccionarios e IF)
    # ------------------------------------------------------
    if opcion == "1":
        print("\n--- NUEVA CAPTURA DE DATOS AGRÍCOLAS ---")

        # Reiniciamos el diccionario para el nuevo registro
        agricultor = {}

        # Solicitud de datos al usuario (Valores para las Keys)
        cultivo_input = input("Tipo de cultivo (ej. Maiz, Trigo, Aguacate): ")
        hectareas_input = input("Cantidad de hectáreas: ")
        hectareas_input = int(hectareas_input) # Conversión a entero
        maquinaria_input = input("Maquinaria agrícola principal (ej. Tractor, Cosechadora): ")
        temporada_input = input("Temporada de cosecha (ej. Primavera, Otono): ")
        ubicacion_input = input("Ubicación del terreno (ej. Norte, Sur, Valle): ")
        fertilizante_input = input("Tipo de fertilizante usado (quimico / organico): ")

        # Procesamos los datos con nuestra función de IA antes de guardarlos
        resultado_ia = analizar_con_ia(cultivo_input, hectareas_input, fertilizante_input)

        # Llenamos el diccionario "agricultor" asignando {Key: Value}
        # Nota: Usamos exactamente los atributos solicitados
        agricultor["Tipo de cultivo"] = cultivo_input
        agricultor["hectáreas"] = hectareas_input
        agricultor["maquinaría agrícola"] = maquinaria_input
        agricultor["temporada de cosecha"] = temporada_input
        agricultor["ubicación del terreno"] = ubicacion_input
        agricultor["fertilizantes usados"] = fertilizante_input
        agricultor["diagnostico_ia"] = resultado_ia # Guardamos el diagnóstico generado

        # Guardamos el diccionario dentro de la lista global
        lista_agricultores.append(agricultor)

        print("\n🤖 AGRO-IA: Registro procesado y optimizado con éxito.")
        print(f"Diagnóstico emitido: {resultado_ia}")

    # ------------------------------------------------------
    # OPCIÓN 2: MOSTRAR TABLA (Llamada a la función)
    # ------------------------------------------------------
    elif opcion == "2":
        mostrar_tabla_agricultores()

    # ------------------------------------------------------
    # OPCIÓN 3: SALIR
    # ------------------------------------------------------
    elif opcion == "3":
        print("\nCerrando Agro-IA... Conexión con los tractores autónomos finalizada.")
        break

    # CONTROL DE ENTRADAS INCORRECTAS
    else:
        print("❌ Opción inválida. Por favor, selecciona 1, 2 o 3.")

    print("\n--------------------------------------------------")
```
